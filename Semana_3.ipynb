```python
import pandas as pd
import pandas as pd
import numpy as np

np.random.seed(42)

n_pacientes = 1000

pacientes = pd.DataFrame({
    "id": np.arange(1, n_pacientes + 1),
    "edad": np.random.randint(18, 90, n_pacientes),
    "peso": np.random.randint(50, 120, n_pacientes),
    "sexo": np.random.choice(["M","F"], n_pacientes),
    "altura_cm": np.random.randint(150, 195, n_pacientes)
})

# -------------------------
# 2. Signos vitales grandes
# cada paciente tiene entre 3 y 10 mediciones
# -------------------------
registros = []

for pid in pacientes["id"]:
    n_med = np.random.randint(3, 11)
    for _ in range(n_med):
        registros.append([
            pid,
            np.random.randint(100, 160),   # presión
            np.random.randint(60, 110),    # fc
            np.random.randint(90, 100),    # spo2
            round(np.random.uniform(36, 39),1)  # temp
        ])

signos = pd.DataFrame(registros, columns=["id","presion","fc","spo2","temp"])
```

```python
# -------------------------
# 2. Merge (unir tablas)
# -------------------------
df = pacientes.merge(signos, on="id",how="right")
```

```python
df
```

```python
# -------------------------
# 3. Variable derivada
# -------------------------
df["riesgo"] = df["presion"] * df["fc"] / 1000
```

```python
df
```

```python
# -------------------------
# 4. Normalización (escala 0–1)
# -------------------------
df["edad_norm"] = (df["edad"] - df["edad"].min()) / (df["edad"].max() - df["edad"].min())
```

```python
df
```

```python
# -------------------------
# 5. GroupBy (resumen por paciente)
# -------------------------
resumen = df.groupby("id").agg({
    "presion": "mean",
    "fc": "mean",
    "riesgo": "mean"
}).reset_index()
```
